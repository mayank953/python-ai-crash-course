## Problem 1: Hypotenuse with `math`

A right triangle has legs of length 3 and 4. Use the `math` module to store the hypotenuse in `hyp`.

```python
import math

hyp = math.sqrt(3**2 + 4**2)
```

```python
assert hyp == 5, 'hyp should be 5.0'
print('✅ Problem 1 passed')
```

## Problem 2: Valid identifiers

From `candidates`, keep only the names Python accepts as variable names. Store them in `valid` (any order).

```python
candidates = ["my_var", "2fast", "_x", "class", "total$", "total2", "user-name", "Name"]
valid = ["my_var", "_x", "total2", "Name"]
```

```python
assert sorted(valid) == ['Name', '_x', 'my_var', 'total2'], 'Remember: no leading digit, no symbols, no keywords.'
print('✅ Problem 2 passed')
```

## Problem 3: Repair the code

`broken` holds a line of code with a syntax error. Put a corrected version in `fixed`.

```python
broken = 'print("Hello, Python"'
fixed = 'print("Hello, Python")'
```

Label each snippet `"syntax"` (Python refuses to start) or `"runtime"` (it starts and then fails).

## Problem 5: Built-in functions

Using only the built-ins `abs`, `round` and `max`, store the largest of `abs(-12)`, `round(7.6)` and `9` in `biggest`.

```python
biggest = max(abs(-12), round(7.6), 9)
```

```python
assert biggest == 12
print('✅ Problem 5 passed')
```
