## Problem 1: Slicing practice

Given `s`, create `first` (first character), `last` (last character), `word1` (the word `Python`), `rev` (the whole string reversed) and `every_other` (every second character).

```python
s = "Python Programming"
first = s[0]
last = s[-1]
word1 = s[:6]
rev = s[::-1]
every_other = s[::2]
```

```python
assert (first, last, word1) == ('P', 'g', 'Python')
assert rev == 'gnimmargorP nohtyP'
assert every_other == 'Pto rgamn'
print('✅ Problem 1 passed')
```

## Problem 2: Casting

`a` and `b` are strings. Store their numeric sum in `sum_num` (an int) and their text join in `joined`.

```python
a, b = "12", "30"
sum_num = int(a) + int(b)
joined = a + b
```

```python
assert sum_num == 42 and isinstance(sum_num, int)
assert joined == '1230'
print('✅ Problem 2 passed')
```

## Problem 3: Truthy or falsy?

Without running `bool()`, fill `answers` with what `bool(value)` returns for each item of `values`. Then check yourself.

## Problem 4: Complex numbers and float precision

For `z = 3 + 4j` store its magnitude in `magnitude` and its conjugate in `conj`. Then set `close` to `True` if `0.1 + 0.2` equals `0.3` when rounded to 2 decimals.

```python
z = 3 + 4j
magnitude = abs(z)
conj = z.conjugate()
close = round(0.1 + 0.2, 2) == 0.3
```

```python
assert magnitude == 5.0
assert conj == 3 - 4j
assert close is True
print('✅ Problem 4 passed')
```

## Problem 5: Palindrome and repetition

Set `is_pal` to `True` if `word` reads the same backwards (ignore case), and `line` to a string of 20 `=` characters.

```python
word = "Madam"
is_pal = word.lower() == word.lower()[::-1]
line = "=" * 20
```

```python
assert is_pal is True
assert line == '=' * 20
print('✅ Problem 5 passed')
```
