## Problem 1: Quotient and remainder

Divide 47 by 5. Store the whole-number result in `quotient`, the leftover in `remainder`, and `2` to the power `10` in `power`.

```python
quotient = 47 // 5
remainder = 47 % 5
power = 2 ** 10
```

```python
assert (quotient, remainder, power) == (9, 2, 1024)
print('✅ Problem 1 passed')
```

## Problem 3: Entry rules

A visitor may enter if they are at least 18 **and** hold a ticket. Store the result in `can_enter`. Also set `has_a` to whether the letter `"a"` is in `"banana"`.

```python
age = 25
has_ticket = True
can_enter = age >= 18 and has_ticket
has_a = "a" in "banana"
```

## Problem 4: `==` versus `is`

For the lists below set `eq` (same contents?), `same` (`a` and `b` are the same object?) and `same_c` (`a` and `c` are the same object?). Predict first, then verify with an expression.

```python
a = [1, 2, 3]
b = [1, 2, 3]
c = a
eq = a == b
same = a is b
same_c = a is c
```

```python
assert (eq, same, same_c) == (True, False, True)
print('✅ Problem 4 passed')
```

## Problem 5: Format a table row

Build `line` with an f-string so the name is left-aligned in 8 characters, followed by `|`, then the marks right-aligned in 8 characters with 2 decimals.

```python
name = "Asha"
marks = 87.456
line = f"{name:<8}|{marks:>8.2f}"
```

```python
assert line == 'Asha    |   87.46', repr(line)
print('✅ Problem 5 passed')
```
