## Problem 1: Grade ladder

Rules: 90 or more is `A`, 80 or more is `B`, 70 or more is `C`, otherwise `D`. Loop over the marks and append each grade letter to the string `result`.

```python
result = ""
for m in [95, 85, 72, 40]:
    if m >= 90:
        result += "A"
    elif m >= 80:
        result += "B"
    elif m >= 70:
        result += "C"
    else:
        result += "D"
```

```python
assert result == 'ABCD'
print('✅ Problem 1 passed')
```

## Problem 2: FizzBuzz

For 1 to 15: print-style logic, but build one string `out` with the items separated by spaces. Multiples of 3 become `Fizz`, of 5 become `Buzz`, of both `FizzBuzz`.

```python
out = ""
for n in range(1, 16):
    if n % 15 == 0:
        item = "FizzBuzz"
    elif n % 3 == 0:
        item = "Fizz"
    elif n % 5 == 0:
        item = "Buzz"
    else:
        item = str(n)
    out += item + " "
out = out.strip()
```

```python
assert out == '1 2 Fizz 4 Buzz Fizz 7 8 Fizz Buzz 11 Fizz 13 14 FizzBuzz'
print('✅ Problem 2 passed')
```

## Problem 3: Digit sum with `while`

Use a `while` loop to add up the digits of `n` (2026 gives 2 + 0 + 2 + 6). Store the sum in `total`.

```python
n = 2026
total = 0
while n > 0:
    total += n % 10
    n //= 10
```

```python
assert total == 10
print('✅ Problem 3 passed')
```

## Problem 4: `continue` and `break`

Add the numbers 1, 2, 3, ... to `total`, but skip multiples of 3 (`continue`) and stop (`break`) as soon as `total` is above 100. Keep the last number you added in `last`.

```python
total = 0
last = None
for n in range(1, 100):
    if n % 3 == 0:
        continue
    total += n
    last = n
    if total > 100:
        break
```

```python
assert (total, last) == (108, 17)
print('✅ Problem 4 passed')
```

## Problem 5: Star triangle

Build `pattern` with nested loops so it holds a triangle of height 4, one row per line: `*`, `**`, `***`, `****` (separate rows with `\n`, no trailing newline).

```python
pattern = ""
for row in range(1, 5):
    for _ in range(row):
        pattern += "*"
    if row < 4:
        pattern += "\n"
```

```python
assert pattern == '*\n**\n***\n****'
print('✅ Problem 5 passed')
```
