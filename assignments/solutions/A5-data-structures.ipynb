## Problem 1: List toolkit

From `nums` create `sorted_desc` (largest first), `top3` (the three largest) and `total` (sum). Do not change `nums` itself.

```python
nums = [5, 3, 8, 1, 9, 2]
sorted_desc = sorted(nums, reverse=True)
top3 = sorted_desc[:3]
total = sum(nums)
```

```python
assert sorted_desc == [9, 8, 5, 3, 2, 1]
assert top3 == [9, 8, 5]
assert total == 28
assert nums == [5, 3, 8, 1, 9, 2]
print('✅ Problem 1 passed')
```

## Problem 2: Comprehension

In one comprehension, build `even_squares`: the squares of the even numbers from 1 to 10.

```python
even_squares = [n * n for n in range(1, 11) if n % 2 == 0]
```

```python
assert even_squares == [4, 16, 36, 64, 100]
print('✅ Problem 2 passed')
```

## Problem 3: Tuples

Swap `a` and `b` using tuple unpacking, then store how many times `2` appears in `t` in `twos`.

```python
a, b = 1, 2
a, b = b, a
t = (1, 2, 2, 3, 2)
twos = t.count(2)
```

```python
assert (a, b) == (2, 1)
assert twos == 3
print('✅ Problem 3 passed')
```

## Problem 4: Set algebra

Find `common` (in both), `only_a` (in `a` but not `b`), `everything` (in either) and `unique_count` (how many distinct values are in `dupes`).

```python
a = {1, 2, 3, 4}
b = {3, 4, 5}
dupes = [1, 1, 2, 3, 3]
common = a & b
only_a = a - b
everything = a | b
unique_count = len(set(dupes))
```

```python
assert common == {3, 4} and only_a == {1, 2}
assert everything == {1, 2, 3, 4, 5}
assert unique_count == 3
print('✅ Problem 4 passed')
```

## Problem 5: Word counter

Build `freq`, a dictionary that maps each word of `sentence` to how many times it appears.

```python
sentence = "to be or not to be"
freq = {}
for word in sentence.split():
    freq[word] = freq.get(word, 0) + 1
```

```python
assert freq == {'to': 2, 'be': 2, 'or': 1, 'not': 1}
print('✅ Problem 5 passed')
```

## Problem 6: String methods and Caesar shift

Clean `text` into `clean` (no outer spaces, lowercase, no comma). Then shift every letter of `word` one step forward in the alphabet into `shifted` using `ord()` and `chr()`.

```python
text = "  Hello, World  "
word = "abc"
clean = text.strip().lower().replace(",", "")
shifted = ""
for ch in word:
    shifted += chr(ord(ch) + 1)
```

```python
assert clean == 'hello world'
assert shifted == 'bcd'
print('✅ Problem 6 passed')
```
