# Abstraction

```python
from abc import ABC,abstractmethod
```

```python
class User(ABC):
    # def __init__
    
    @abstractmethod
    def login(self):
        pass

    def logout(self):
        print("logout")

    @abstractmethod
    def auth(self):
        pass
```

```python
class Buyer(User):

    def login(self):
        if(self.auth):
            print("Logging in User")
            
    def checkObject(self,link):
        print(link)

    def auth(self):
        return True
```

```python
class Seller(User):

    def login(self):
        if(self.auth):
            print("Logging in Seller")
    def checkObject(self,link):
        print(link)

    def auth(self):
        print("OTP")
        return True
```

```python
b1 = Buyer()
```

```python
b1.auth()
```

```python
from abc import ABC,abstractmethod

class Animal(ABC):
    @abstractmethod
    def speak(self):
        pass
```

```python
class Dog(Animal):
    def speak2(self):
        print ("bark")
```

```python
d1 = Dog()
```

```python
d1.speak()
```
