# Break Statement


```python
start = -2
end = 2
num = 10
```

```python
while (start <= end):
    if(start == 0):
        break
    print(num/start)
    start = start + 1
    print(" ans printed ")

print("Done")

```

# Continue

```python
start = -2
end = 2
num = 10
```

```python
while (start <= end):
    if(start == 0):
        start = start + 1
        continue
    print(num/start)
    start = start + 1
    print(" ans printed ")

print("Done")

```
