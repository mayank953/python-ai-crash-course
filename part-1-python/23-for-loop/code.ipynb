# For Loop

```python
name = "Mayank"
```

```python
for ch in name:
    print(ch)
```

```python
l1 = [1,2,3,4,5]

for i in l1:
    if(i ==3):
        i = i+1
        continue
    print(i)
    
```

```python
start = 0
while(start < len(l1)):
    print(l1[start])
    start = start + 1
```
