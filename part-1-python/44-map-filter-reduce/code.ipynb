# Map Function

```python
def square(x):
    return x*x
```

```python
def ApplyOnList(func,l1):
    l2 = []
    for i in l1:
        l2.append(func(i))
    return l2
```

```python
tuple(map(square,(1,2,3,4,5)))
```

```python
tuple(map(lambda x : x + 5,(1,2,3,4,5)))
```

# Filter

```python
def FilterMyList(func,l1):
    l1Modified = []
    for i in l1:
        if(func(i)):
            l1Modified.append(i)

    return l1Modified
```

```python
def isEven(num):
    return num%2==0
```

```python
FilterMyList(isEven,[1,2,3,4,5])
```

```python
f1 = list(filter(isEven,[1,2,3,4,5]))
```

```python
list(f1)
```

```python
list(filter(lambda x : x*x , [1,2,5,10,20,100,150,9]))
```

```python
from functools import reduce
```

```python
reduce(lambda a,b: a*b ,[1,2,3,4,5])
```
