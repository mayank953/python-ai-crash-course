# Logical Operators

```python
a = 6
b = 20
c = 50
```

```python

# Using 'and' operator
print(a < b and b < c)  # True because both conditions are true

# True And True
```

```python
# Using 'or' operator
print(a > b or b < c)   # True because one of the conditions is true

# False or True
```

```python

# Using 'not' operator
print(not (a > b))      # True because the condition is false and 'not' negates it
```
